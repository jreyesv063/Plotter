```python
import pandas as pd

sample = "SignalTau_1000"
# Seleccionar las columnas que te interesan
signal_cols = [c for c in events.columns if c.startswith(f"{sample}")]

# Asignar nombres más cómodos
down_col = [c for c in signal_cols if "Down" in c][0]
nom_col  = [c for c in signal_cols if c.endswith("GeV") and "Up" not in c and "Down" not in c][0]
up_col   = [c for c in signal_cols if "Up" in c][0]

# Calcular las diferencias porcentuales
events[f"{sample}GeV Up (%)"] = ((events[up_col] - events[nom_col]) / events[nom_col]) * 100
events[f"{sample}GeV Down (%)"] = ((events[down_col] - events[nom_col]) / events[nom_col]) * 100

# Mostrar las columnas relevantes
pd.set_option('display.max_columns', None)
events[[down_col, nom_col, up_col, f"{sample}GeV Up (%)", f"{sample}GeV Down (%)"]]
```
